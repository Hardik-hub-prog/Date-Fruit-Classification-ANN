### Deep Learning

```python
import os
os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"
```

```python
import torch  # short Pytorch - torch

print(torch.__version__)
print(torch.cuda.is_available())
```

```python
# Convert data into Tensors

import torch.nn as nn

X_train_tensor = torch.tensor(X_train_scale , dtype = torch.float32)
X_test_tensor = torch.tensor(X_test_scale , dtype = torch.float32)
y_train_tensor = torch.tensor(y_train , dtype = torch.long )
y_test_tensor = torch.tensor(y_test , dtype = torch.long)
```

```python
# Loss fnx for classification problem Cross entropy , target value expect long format that's why - torch.long
```

```python
type(y_train)
```

```python
# Tensor Data set - To define Dataset, access from Memory(RAM)
```

```python
from torch.utils.data import TensorDataset , DataLoader

#train_dataset = TensorDataset(in_features , op_features)
#instance create for class TensorDataset

train_dataset = TensorDataset(X_train_tensor , y_train_tensor)
test_dataset = TensorDataset(X_test_tensor , y_test_tensor)
```

```python
# Data Loader - To define how the data will be loaded for traning
```

```python
train_loader = DataLoader(train_dataset , batch_size =32 , shuffle = True)
test_loader = DataLoader(test_dataset , batch_size = 32)
```

```python
X_train.shape[1]
```

### Build Model

```python
class ANN(nn.Module):
    def __init__(self):
        super(ANN , self).__init__()

        self.model = nn.Sequential(

            # 1st hidden layer
            # nn.AF(in_feature , op_feature),

            nn.Linear(X_train.shape[1] , 64),
            nn.ReLU(),

            # 2nd hidden layer

            nn.Linear(64,64),
            nn.ReLU(),

            #O/P Layer

            nn.Linear(64,7)
            # Softmax() - by deafult cross entropy - Cost fnx - there apply
            
            
        )

    def forward(self , x):
        return self.model(x)

        
```

```python
import torch.optim as optim

model = ANN()

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters())


```

### Train Model

```python
epochs = 100

for epoch in range(epochs):
    model.train()
    running_loss = 0.0

    for xb , yb in train_loader:

        optimizer.zero_grad() # for each batch fresh gradient compute, dont accumualte value of graident of previous batch
        
        output = model(xb)  #forward propgation... predicted value for this batch
        loss = criterion(output, yb)  # Loss compute
        loss.backward() # Gradient compute automatically
        optimizer.step() # learn params update

        running_loss += loss  # Total running loss cal for all batches(23) in 1 epoch

    train_loss = running_loss/len(train_loader)  # Avg loss per batch for 1 epoch
        
    print(f"epoch = {epoch}/{epochs} , loss = {train_loss}")    
        
        
        
    
    

```

### Evaluation

```python
model.eval()

total = 0
correct = 0  # correct pred

with torch.no_grad():

    for xb , yb in test_loader:
        output = model(xb) #Forward Propgation , pred value
        _, pred = torch.max(output,1) # torch.max - return max value and index
        # here pred = index
        # o/p  - 7 values [1, 0.6, 1.4 , 0.5 , -0.2 , 1.2, 0.1 ]
        #( max value = 1.4 , index = 2)

        correct += (pred == yb).sum().item()  # sum store for all batches

        

    



```
